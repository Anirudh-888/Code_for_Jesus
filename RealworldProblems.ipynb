# A shopkeeper enters the names, quantities, and prices of 3 items. Calculate the total bill and display it neatly. If the bill is above ₹1,000, give a 10% discount.

```python
item1=input("Enter the name of item 1")
quantity1=int(input(f"Enter the number of {item1}"))
price1=int(input(f"Enter the price of 1 {item1}"))
item2=input("Enter the name of item 2")
quantity2=int(input(f"Enter the number of {item2}"))
price2=int(input(f"Enter the price of 1 {item2}"))
item3=input("Enter the name of item 3")
quantity3=int(input(f"Enter the number of {item3}"))
price3=int(input(f"Enter the price of 1 {item3}"))
print("The final bill")
cost1=quantity1*price1
cost2=quantity2*price2
cost3=quantity3*price3
totalcost=cost1+cost2+cost3
print("The total bill amount is",totalcost)
if(totalcost>1000):
    discount=(10/100)*totalcost
    print("The final bill amount is",totalcost-discount)
```

# Replicate a banking System and then write the code for the banking system with deposit and withdrawl and the pin attempt should be allowed only 3 times


```python
balance = 100000000000
pin=input("Enter the pin")
attempt=0
while(attempt<3):
    if pin =="110405":
        choice=input("Enter the choice 1 for Deposit 2 for Withdrawl 3 for balance")
        if choice=="1":
            amount=int(input("Enter the amount you want to deposit"))
            balance = balance+amount
            print("The current balance after deposit is",balance)
        elif choice=="2":
            if(amount<balance):
                amount=int(input("Enter the amount you want to withdraw"))
                balance=balance-amount
                print("The current balance after withdrawl is",balance)
            else:
                print("No sufficient balance")
        elif choice =="3":
            print("The current balance in your account is",balance)
        else:
            print("Enter the proper choice")
    else:
        attempt=attempt+1
        pin=input("Enter the proper pin")
        print("The chances left are",3-attempt)
```
