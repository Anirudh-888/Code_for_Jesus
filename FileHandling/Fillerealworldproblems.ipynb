```python
with open("studentrecord.txt","a+") as file:
    print("The details in the student file")
    student_name=input("Enter the name of the student")
    marks1=(input("Enter the marks 1 of the student"))
    marks2=(input("Enter the marks 2 of the student"))
    marks3=(input("Enter the marks 3 of the student"))
    total_marks=float(marks1)+float(marks2)+float(marks3)
    data=(f"Name:{student_name},Marks 1:{marks1}, Marks 2:{marks2}, Marks 3:{marks3},Total Marks:{total_marks} \n")
    file.write(data)
    print("Reading the contents of the file")
    file.seek(0)
    content=file.read()
    print("The student list and their marks\n",content)
```
