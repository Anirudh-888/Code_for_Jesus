# Reading the file

# Path of the file: The location in which is present in the computer storage/ROM
* *File in Python can be accessed with its path*
# Syntax to open the file
* *file_variable = open("filepath","modeofaccess",encoding)*
# Note: After opening the file in the read mode we have to close it mandatorily
* *Syntax to close the file: file_variable.close()*

```python
fn="Psalms23.txt"
file=open(fn,"r",encoding="utf-8")
content=file.read()
print(content)
file.close()
```

# Types of file reading:
* *1. read() reads the entire file*
* *2. readline() reads the file line by line*
* *3. readlines() reads all the lines into a list*
## All the codes are below

* *1.read() Reading the entire files*

```python
filename="sample.txt"
file=open(filename,"r",encoding="utf-8")
text=file.read()
print(text)
file.close()
```

* *2. readline() reads the file line by line*

```python
filename="sample.txt"
file=open(filename,"r",encoding="utf-8")
print("The first line in the file")
text=file.readline()
print(text)
file.close()
```

* *3 readlines() reads the entire file into list*

```python
filename="sample.txt"
file=open(filename,"r",encoding="utf-8")
text=file.readlines()
print(f"The text stored in the form of list:{text}")
j=1
for i in text:
    print(f"The line {j} in the file is {i}")
    j+=1
file.close()
```

# Write operation in files:


```python
filename="sample.txt"
file=open(filename,"r")
text=file.read()
print("The initial content in the file:",text)
with open(filename,"w") as file:
    file.write("Hello Guys this is the change in the file")
    file.close()
file=open(filename,"r")
text1=file.read()
print("File text after changing the content of the file",text1)
```
