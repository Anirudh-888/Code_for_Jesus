# Types of file reading:
* *1. read() reads the entire file*
* *2. readline() reads the file line by line*
* *3. readlines() reads all the lines into a list*
## All the codes are below

* *1.read() Reading the entire files*

```python
filename="sample.txt"
file=open(filename,"r",encoding="utf-8")
text=file.read()
print(text)
file.close()
```

* *2. readline() reads the file line by line*

```python
filename="sample.txt"
file=open(filename,"r",encoding="utf-8")
print("The first line in the file")
text=file.readline()
print(text)
file.close()
```

* *3 readlines() reads the entire file into list*

```python
filename="sample.txt"
file=open(filename,"r",encoding="utf-8")
text=file.readlines()
print(f"The text stored in the form of list:{text}")
j=1
for i in text:
    print(f"The line {j} in the file is {i}")
    j+=1
file.close()
```

# Write operation in files:


```python
filename="sample.txt"
file=open(filename,"r")
text=file.read()
print("The initial content in the file:",text)
with open(filename,"w") as file:
    file.write("Hello Guys this is the change in the file")
    file.close()
file=open(filename,"r")
text1=file.read()
print("File text after changing the content of the file",text1)
```

# Appending the text in the file:
* ## This will not change the file like that of the write mode it will add the content at the end of the file.
# mode of access is "a"

```python
fname="appendexample.txt"
appendfile=open(fname,"r",encoding="utf-8")
text=appendfile.read()
print(f"The initial content in the file {text}")
with open(fname,"a") as appendfile:
    appendfile.write("\n This is written by your Anirudh Mentor")
    appendfile.close()
appendfile=open(fname,"r",encoding="utf-8")
updatedtext=appendfile.read()
print(f"The updated content in the file {fname} is {updatedtext}")
```

```python
fname="appendexample.txt"
appendfile=open(fname,"rb")
text=appendfile.read()
print(f"The initial content in the file {text}")
print(list(text))
for byte in text:
    print(format(byte,"08b"))
```

# Reading the image as a binary file:

```python
imagepath="binaryimage.jpg"
with open(imagepath,"rb") as image:
    bimage=image.read()
print(type(bimage))
print(len(bimage))
print(list(bimage[:20]))
for byte in bimage:
    print(format(byte,"08b"))
```
