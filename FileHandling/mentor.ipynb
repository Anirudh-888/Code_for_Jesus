```python
'''Personal Expense tracker: Records the daily expenses in a file.
Store the date, expense category and amount
Display all expenses
Calculate total monthly spending
Find expenses belonging to a particular category
'''
print("Welcome to the STC Tracker")
with open("data.txt","a")as file:
    content=file.read()
    print("The previous tracking data present in the file",content)
    date=int(input("Enter the date of the transaction"))
    month=int(input("Enter the month of the transaction"))
    year=int(input("Enter the year of the transaction"))
    fdate=(str(date)+str(month)+str(year))
    print("Enter the choice for the category of the transaction")
    print("1. Grocery 2. Transport 3.Entertainment 4.Medical Expenses")
    choice=input()
    if choice=="1":
             item=input("Enter the name of the transaction")
             Grocery=[(fdate,item)]
    
    elif choice=="2":
             item=input("Enter the name of the transaction")
             Transport=[(fdate,item)]
    
    elif choice=="3":
             item=input("Enter the name of the transaction")
             Entertainment=[(fdate,item)]

    elif choice=="4":
             item=input("Enter the name of the transaction")
             MedExpense=[(fdate,item)]
    else:
             print("Enter the choice properly")
    print(Grocery)
    print(Transport)
    print(Entertainment)
    print(MedExpense)
```
