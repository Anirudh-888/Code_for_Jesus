# Writing in the file:

```python
file=open("sample.txt","r")
initialtext=file.read()
print(initialtext)
file.close()
with open("C:\Code_for_Immanuel\Code_for_Jesus\FileHandling\sample.txt","w+") as file:
    file.write("But by the grace and the training from the Lord she didn't fight")
    print("The file content after writing")
file=open("sample.txt","r")
finaltext=file.read()
print(finaltext)
file.close()
```

```python
filepath="sample.txt"
file=open(filepath,"r")
it=file.read()
print("The initial content in the file is",it)
file.close()
with open(filepath,"a") as file:
    file.write("She didn't speak a bad word and was surrendered to the Lord")
file=open(filepath,"r")
ft=file.read()
print("The content in the file after appending",ft)
file.close()
```
