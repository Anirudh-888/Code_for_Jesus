```python
file=open("studentdetails.txt","a+")
content=file.read()
print(content)
student_name=input("Enter the name of the student")
marks1=float(input("Enter marks 1 of the student"))
marks2=float(input("Enter marks 2 of the student"))
marks3=float(input("Enter marks 3 of the student"))
totalmarks=marks1+marks2+marks3
data=(f"Name: {student_name} Marks1 :{marks1} Marks2:{marks2} Marks3:{marks3}. Total Marks of the student: {totalmarks}\n")
file.write(data)
file.seek(0)
details=file.read()
print(details)
```
