# Square Patterns


```python
n=int(input("Enter the number till the pattern is required"))
for i in range(n):
    for j in range(n):
        print("Anirudh",end=" ")
    print()
```

```python
n=int(input("Enter the number of rows requried for the pattern"))
for i in range(n):
    for j in range(i+1):
        print("Anirudh",end=" ")
    print()
```

```python
n=int(input("Enter the number of rows requried for the pattern"))
for i in range(n):
    for j in range(n-i):
        print("Anirudh",end=" ")
    print()
```
