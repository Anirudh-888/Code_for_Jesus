# Write a Python program to accept a string and a character from the user. Count and display how many times the given character occurs in the string. Use string traversal and the concepts of indexing/length or the string count() function taught in the chapter.

```python
string=input("Enter the string for the question")
char=input("Enter the character for the question")
print("The number of times the character is repeating in the string",string.count(char))
# The same code using for loop
count=0
for ch in string:
    if ch==char:
        count=count+1
    else:
        count=count+0
print("The number of times the character is in the string is",count)
```

# Write a Python program to accept a string from the user and display the string in reverse order. Use string indexing/slicing concepts from the chapter.

```python
string=input("Enter the string that is to be reversed")
for i in range(-1,-len(string)-1,-1):
    print(string[i],end="")
```

# Check for Palindrome
# Write a Python program to accept a string and check whether it is a palindrome. A palindrome reads the same from left to right and right to left. Display an appropriate message for the result.

```python
sample=input("Enter the string to check whether it is a palindrome or not")
status=False
for i in range(len(sample)):
    if sample[i] == sample[len(sample)-1-i]:
        status=True
    else:
        status=False
if status is True:
    print(f"The entered string {sample} is a palindrome")
else:
    print(f"The entered string {sample} is not a palindrome")
```

# Count Vowels
# Write a Python program to accept a string and count the number of vowels (a, e, i, o, u) present in it.Traverse the string and use the membership operator in to check whether each character is a vowel

```python
checking_string=input("Enter the string to count the number of vowels")
count=0
for ch in checking_string:
    if ch in "aeiouAEIOU":
        count+=1
    else:
        count+=0
print(f"The number of vowels in the entered string{checking_string} is {count}")
```
