# 1. Swapping two numbers:

```python
a=input("Enter the first number to be swapped")
b=input("Enter the second number to be swapped")
print("before swapping")
print(a,b)
temp=a
a=b
b=temp
print("After swapping")
print(a,b)
```

# 2. Arithematic operators


```python
num1=float(input("Enter the first number"))
num2=float(input("Enter the second number"))
print("Addition",num1+num2)
print("Subtraction",num1-num2)
print("Multiplication",num1*num2)
print("Division",num1/num2)
print("Modulus",num1%num2)
print("Floor division",num1//num2)
print("Power",num1**num2)
```

# 3. Area and the perimeter of the rectangle

```python
length=float(input("Enter the length of the rectangle"))
breadth=float(input("Enter the breadth of the rectangle"))
print("The perimeter of the rectangle is",2*(length+breadth))
print("The area of the rectangle is",length*breadth)
```

# 4. Simple Interest

```python
principle=float(input("Enter the amount borrowed"))
rate=float(input("Enter the rate of interest"))
time=float(input("Enter the time for returning the money"))
print("The simple interest is",(principle*rate*time)/100)
print("The total amount to be returned is",principle+(principle*rate*time)/100)
```

# 5.Largest of three numbers

```python
num1=float(input("Enter the first number"))
num2=float(input("Enter the second number"))
num3=float(input("Enter the third number"))
if num1>num2:
    if num1>num3:
        print("The largest number is",num1)
    else:
        print("The largest number is",num3)
if num2>num1:
    if num2>num3:
        print("The largest number is",num2)
    else:
        print("The largest number is",num3)
```

# 7.Minimum and maximum of 5 numbers

```python
num=[]
print("Enter the numbers")
for i in range(6):
    a=int(input(f"Enter the {i+1} number"))
    num.append(a)
print("The maximum of the 5 numbers",max(num))
print("The minimum of the 5 numbers",min(num))
    
    
```

# 8. Grade Card


```python
marks=float(input("Enter the percentage"))
if marks>=90:
    print("GRADE is A")
elif marks<90 and marks >=80:
    print("GRADE is B")
elif marks<80 and marks >=70:
    print("GRADE is C")
elif marks<70 and marks >=60:
    print("GRADE is D")
elif marks<60 and marks >=50:
    print("GRADE is E")
else:
    print("Enter proper percentage or you are failed in the exams")
```

# 9. Printing the Multiplication tables:

# 10. Sum of the digits of a number

```python
sum=0
number=input("Enter the number for the sum of its digits")
for i in number:
    sum=sum+int(i)
print(f"The sum of the digits of the number {number} is:{sum}")
```

# 11. Palindrome:

```python
string=input("Enter the string to check whether it is a palindrome")
for i in range(len(string)-1):
    if string[i]==string[-1-i]:
        valid=True
    else:
        valid=False
if valid:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```
