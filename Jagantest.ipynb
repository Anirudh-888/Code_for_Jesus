```python
#1
count=0
str=input("enter the string: ")
chr=input("enter the character: ")
for i in str:
    if chr in i:
        count+=1
    else:
        count+=0
print(count)
#2
str2=input("enter the word: ")
for j in range(-1,-len(str2)-1,-1):
    print(str2[j],end="")
#3
str3=input("enter the word: ")
status=False
for k in range(len(str3)):
    if str3[k]==str3[len(str3)-1-k]:
        status=True
    else:
        status=False
if status==True:
    print(f"the word {str3} is a palindrome")
else:
    print(f"the word {str3} is not a palindrome")
#4
count1=0
str2=input("enter the word to count vowels: ")
for j in str2:
    if j=="a" or j=="e" or j=="i" or j=="o" or j=="u":
        count1+=1
    else:
        count1+=0
print("the no of vowels in",str2,"is",count1)
#5
str3=input("enter the sentence: ")
for k in str3:
    if k==" ":
        k="_"
    print(k,end="")
print(str3.replace(" ","_"))
```
