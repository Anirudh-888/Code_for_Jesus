# Program 8-2 Write a program with a user define function with string as a parameter which replaces all vowels in the string with '*'.

```python
def voweltostar(name):
    for i in name:
        if i=="a" or i=="e" or i=="i" or i=="o" or i=="u":
            i="*"
        print(i,end="")
voweltostar("Jagana Mohane Reddyi")
```

# Write a program to input a string fromthe user and print it in the reverse order without creating a new string.

```python
string=input("Enter the string that should be reversed")
for i in range(-1,-len(string)-1,-1):
    print(string[i],end="")
```
