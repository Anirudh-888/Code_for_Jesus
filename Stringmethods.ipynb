# String Operations
* *Concatenation: Addition of two strings using +*

```python
string="Anirudh"
string2="Jesus Christ "
string2+string
```

* *Repetition: Repeating of strings multiple times*

```python
string="Anirudh"
print(string*10)
for i in range(0,10):
    print(string)
```

# Membership
* *It will tell us if this character is in the string*
* *It is case sensitive*

```python
string="Anirudh"
print('A' in string)
print("B" in string)
print("ANIRUDH" in string)
```

```python
string="Anirudh"
print(len(string))
print(string.title())
print(string.upper())
print(string.lower())
print(string.isupper())
print(string.islower())
```
